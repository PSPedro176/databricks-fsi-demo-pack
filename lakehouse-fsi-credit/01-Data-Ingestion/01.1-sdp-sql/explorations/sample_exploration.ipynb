### Example Exploratory Notebook

Use this notebook to explore the data generated by the pipeline in your preferred programming language.

**Note**: This notebook is not executed as part of the pipeline.

```sql
%python
catalog = "${var.catalog}"
schema = dbName = db = "fsi_credit"
volume_name = "credit_raw_data"
```

```sql
%python
sdf = spark.read.json(f"/Volumes/{catalog}/{db}/{volume_name}/credit_bureau")
display(sdf)
```

```sql
%python
sdf = spark.read.format("csv").option("header", True).load(f"/Volumes/{catalog}/{db}/{volume_name}/internalbanking/customer")
display(sdf)
```

```sql
%python
sdf = spark.read.format("csv").option("header", True).load(f"/Volumes/{catalog}/{db}/{volume_name}/internalbanking/relationship")
display(sdf)
```

```sql
%python
sdf = spark.read.format("csv").option("header", True).load(f"/Volumes/{catalog}/{db}/{volume_name}/internalbanking/account")
display(sdf)
```
