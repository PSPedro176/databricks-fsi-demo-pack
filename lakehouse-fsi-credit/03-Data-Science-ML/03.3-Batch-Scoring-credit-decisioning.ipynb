````python
import mlflow
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# --- Configuração ---
catalog = "${var.catalog}"
db = "fsi_credit"
model_name = "dbdemos_fsi_credit_decisioning"
table_name = f"{catalog}.{db}.credit_decisioning_features"
model_uri = f"models:/{catalog}.{db}.{model_name}@prod"

# --- Carregar modelo sklearn diretamente ---
mlflow.set_registry_uri('databricks-uc')
sklearn_model = mlflow.sklearn.load_model(model_uri)

# --- Patch para compatibilidade sklearn 1.3 -> 1.6 ---
for tree in sklearn_model.estimators_:
    if not hasattr(tree, 'monotonic_cst'):
        tree.monotonic_cst = None

# --- Extrair features do schema do modelo ---
pyfunc_model = mlflow.pyfunc.load_model(model_uri)
features = pyfunc_model.metadata.get_input_schema().input_names()
print(f"{len(features)} features")

# --- Ler tabela como pandas DataFrame ---
pdf = spark.table(table_name).toPandas()
print(f"Shape: {pdf.shape}")

# --- Preencher NAs com 0 ---
pdf[features] = pdf[features].fillna(0)

# --- Ajustar tipos conforme schema do modelo ---
type_map = {"integer": np.int32, "long": np.int64, "double": np.float64}
schema = pyfunc_model.metadata.get_input_schema()
for col_spec in schema.inputs:
    t = type_map.get(col_spec.type.name)
    if t and col_spec.name in pdf.columns:
        pdf[col_spec.name] = pdf[col_spec.name].astype(t)

# --- Scoring via predict_proba ---
proba = RandomForestClassifier.predict_proba(sklearn_model, pdf[features])
pdf["prediction"] = sklearn_model.classes_[np.argmax(proba, axis=1)].astype(float)

print(f"Prediction distribution:\n{pdf['prediction'].value_counts()}")

# --- Exibir resultado ---
underbanked_df = spark.createDataFrame(pdf)
display(underbanked_df)
````


In the scored data frame above, we have essentially created an end-to-end process to predict credit worthiness for any customer, regardless of whether the customer has an existing bank account. We have a binary prediction which captures this and incorporates all the intellience from Databricks AutoML and curated features from our feature store.
